## Data analysis

### Subtask:
Perform univariate and multivariate analysis on the cleaned dataset.


**Reasoning**:
Start with univariate analysis by examining the distribution of numerical variables like 'Purchase Price', 'Credit Card', and 'CC Security Code' using histograms and boxplots.



```python
import matplotlib.pyplot as plt

# Univariate analysis for numerical variables
numerical_variables = ['Purchase Price', 'Credit Card', 'CC Security Code']
for variable in numerical_variables:
  plt.figure(figsize=(8, 6))
  plt.hist(df[variable], bins=20)
  plt.xlabel(variable)
  plt.ylabel('Frequency')
  plt.title(f'Histogram of {variable}')
  plt.show()

  plt.figure(figsize=(8, 6))
  plt.boxplot(df[variable])
  plt.ylabel(variable)
  plt.title(f'Boxplot of {variable}')
  plt.show()
```

**Reasoning**:
Continue with univariate analysis by examining the frequency of categorical variables like 'Language', 'Job', 'Browser Info', and 'Company'.



```python
# Univariate analysis for categorical variables
categorical_variables = ['Language', 'Job', 'Browser Info', 'Company']
for variable in categorical_variables:
  value_counts = df[variable].value_counts()
  print(f"Frequency of {variable}:\n{value_counts}\n")
  plt.figure(figsize=(12, 6))
  plt.bar(value_counts.index, value_counts.values)
  plt.xlabel(variable)
  plt.ylabel('Frequency')
  plt.title(f'Bar Chart of {variable} Frequency')
  plt.xticks(rotation=90)
  plt.show()
```

**Reasoning**:
Proceed with multivariate analysis by identifying potential relationships between pairs of variables using scatter plots or correlation matrices. Focus on the relationship between 'Purchase Price' and 'Time on Site', if available in the dataset.



```python
# Multivariate analysis
# Check if 'Time on Site' is available in the dataframe.
if 'Time on Site' in df.columns:
  plt.figure(figsize=(8, 6))
  plt.scatter(df['Time on Site'], df['Purchase Price'])
  plt.xlabel('Time on Site')
  plt.ylabel('Purchase Price')
  plt.title('Scatter Plot of Purchase Price vs Time on Site')
  plt.show()

  correlation = df['Time on Site'].corr(df['Purchase Price'])
  print(f"Correlation between Time on Site and Purchase Price: {correlation}")

# If 'Time on Site' is not available, you can analyze other variable pairs.
# For example, you can analyze the correlation between 'Purchase Price' and 'Age'
# if 'Age' in df.columns:
#   plt.figure(figsize=(8, 6))
#   plt.scatter(df['Age'], df['Purchase Price'])
#   plt.xlabel('Age')
#   plt.ylabel('Purchase Price')
#   plt.title('Scatter Plot of Purchase Price vs Age')
#   plt.show()

#   correlation = df['Age'].corr(df['Purchase Price'])
#   print(f"Correlation between Age and Purchase Price: {correlation}")
```

**Reasoning**:
Continue the multivariate analysis by exploring the relationship between categorical variables and numerical variables using group-by operations and visualizations like box plots. For instance, analyze the average 'Purchase Price' for different 'Languages' or 'Jobs'.



```python
# Multivariate analysis: Categorical vs Numerical
# Analyze average 'Purchase Price' for different 'Languages'
plt.figure(figsize=(12, 6))
df.boxplot(column='Purchase Price', by='Language')
plt.title('Boxplot of Purchase Price by Language')
plt.suptitle('')  # Remove the default suptitle
plt.ylabel('Purchase Price')
plt.xticks(rotation=90)
plt.show()

# Analyze average 'Purchase Price' for different 'Jobs' (limit to top 10 jobs for better visualization)
top_10_jobs = df['Job'].value_counts().head(10).index
plt.figure(figsize=(12, 6))
df[df['Job'].isin(top_10_jobs)].boxplot(column='Purchase Price', by='Job')
plt.title('Boxplot of Purchase Price by Job (Top 10)')
plt.suptitle('')  # Remove the default suptitle
plt.ylabel('Purchase Price')
plt.xticks(rotation=90)
plt.show()
```

**Reasoning**:
Analyze the interaction between multiple numerical variables, for example, identify trends or patterns in 'Purchase Price' when considering both 'Time on Site' and 'Age' if these variables are available.



```python
# Multivariate analysis: Interaction between multiple numerical variables
if 'Time on Site' in df.columns and 'Age' in df.columns:
  from mpl_toolkits.mplot3d import Axes3D

  fig = plt.figure(figsize=(10, 8))
  ax = fig.add_subplot(111, projection='3d')
  ax.scatter(df['Time on Site'], df['Age'], df['Purchase Price'])
  ax.set_xlabel('Time on Site')
  ax.set_ylabel('Age')
  ax.set_zlabel('Purchase Price')
  plt.title('Scatter Plot of Purchase Price vs Time on Site and Age')
  plt.show()

# If 'Time on Site' or 'Age' is not available, you can analyze other combinations of numerical variables.
# For example, you can analyze the relationship between 'Purchase Price' and 'Credit Card' and 'CC Security Code'.
# plt.figure(figsize=(8, 6))
# plt.scatter(df['Credit Card'], df['Purchase Price'], c=df['CC Security Code'])
# plt.xlabel('Credit Card')
# plt.ylabel('Purchase Price')
# plt.title('Scatter Plot of Purchase Price vs Credit Card (Color by CC Security Code)')
# plt.colorbar()
# plt.show()
```

## Data visualization

### Subtask:
Create visualizations to illustrate the insights gained from the data analysis.


**Reasoning**:
Create visualizations based on the data analysis results, such as histograms, box plots, bar charts, scatter plots, and 3D scatter plots.



```python
import matplotlib.pyplot as plt

# 1. Distribution of Purchase Price
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.hist(df['Purchase Price'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Purchase Price')
plt.ylabel('Frequency')
plt.title('Histogram of Purchase Price')

plt.subplot(1, 2, 2)
plt.boxplot(df['Purchase Price'], vert=True, patch_artist=True, boxprops=dict(facecolor='lightgreen'))
plt.ylabel('Purchase Price')
plt.title('Box Plot of Purchase Price')

plt.tight_layout()
plt.show()


# 2. Frequency of Languages
language_counts = df['Language'].value_counts()
plt.figure(figsize=(12, 6))
plt.bar(language_counts.index, language_counts.values, color='coral')
plt.xlabel('Language')
plt.ylabel('Frequency')
plt.title('Frequency of Languages Used by Customers')
plt.xticks(rotation=45, ha='right')
plt.show()


# 4. Purchase Price by Job
top_10_jobs = df['Job'].value_counts().head(10).index
plt.figure(figsize=(12, 6))
df[df['Job'].isin(top_10_jobs)].boxplot(column='Purchase Price', by='Job', vert=True, patch_artist=True, boxprops=dict(facecolor='lightblue'))
plt.title('Distribution of Purchase Price Across Top 10 Jobs')
plt.ylabel('Purchase Price')
plt.xlabel('Job')
plt.xticks(rotation=45, ha='right')
plt.suptitle('')  # Remove default subplot title
plt.tight_layout()
plt.show()


# 5. Purchase Price by Language
plt.figure(figsize=(12, 6))
df.boxplot(column='Purchase Price', by='Language', vert=True, patch_artist=True, boxprops=dict(facecolor='lightcoral'))
plt.title('Distribution of Purchase Price Across Languages')
plt.ylabel('Purchase Price')
plt.xlabel('Language')
plt.xticks(rotation=45, ha='right')
plt.suptitle('')  # Remove default subplot title
plt.tight_layout()
plt.show()


# Check for 'Time on Site' and 'Age' columns
if 'Time on Site' in df.columns:
    # 3. Relationship between Purchase Price and Time on Site
    plt.figure(figsize=(8, 6))
    plt.scatter(df['Time on Site'], df['Purchase Price'], color='purple')
    plt.xlabel('Time on Site')
    plt.ylabel('Purchase Price')
    plt.title('Relationship Between Purchase Price and Time on Site')
    plt.show()

if 'Time on Site' in df.columns and 'Age' in df.columns:
    # 6. Interaction between Purchase Price, Time on Site, and Age
    from mpl_toolkits.mplot3d import Axes3D
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(df['Time on Site'], df['Age'], df['Purchase Price'], color='green')
    ax.set_xlabel('Time on Site')
    ax.set_ylabel('Age')
    ax.set_zlabel('Purchase Price')
    plt.title('Interaction Between Purchase Price, Time on Site, and Age')
    plt.show()
```
