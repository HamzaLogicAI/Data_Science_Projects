# Mastering Data Analysis: A Hands-On Warm-Up Exercise with Pandas & Seaborn

## Unlock Insights with Data Exploration & Visualization

### Importing the Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Creating a Sample Dataset

```python
dict1 = {
    'Name': ["Ahmed", "Abdullah", "Alishba", "Fatima", "Adnan", "Ravi", "Ayesha", "Tanveer"],
    'Gender': ["Male", "Male", "Female", "Female", "Male", "Male", "Female", "Male"],
    'Marks': [89, 77, 88, 90, 67, 86, 91, 79]
}
df = pd.DataFrame(dict1)
```

### Display the first few rows

```python
print("Dataset Preview:")
display(df.head())
```

### Display the last few rows

```python
print("\nLast Few Rows:")
display(df.tail())
```

### Shape of the Dataset

```python
print("\nDataset Shape: Rows -", df.shape[0], "| Columns -", df.shape[1])
```

### Checking for Missing Values

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

### Summary Statistics

```python
print("\nSummary Statistics:")
print(df.describe(include='all'))
```

### Visualizing Gender Distribution

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Gender', palette='coolwarm')
plt.title("Gender Distribution in Dataset")
plt.show()
```

### Visualizing Marks Distribution

```python
plt.figure(figsize=(6, 4))
sns.histplot(df['Marks'], bins=5, kde=True, color='blue')
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```
